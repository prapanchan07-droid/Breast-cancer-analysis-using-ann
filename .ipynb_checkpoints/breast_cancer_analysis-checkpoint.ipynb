# Breast Tumor Classification — Complete ML Workflow

**Task:** Binary classification of breast tumors as Benign (B) or Malignant (M).

> Educational demonstration only; not a medical diagnostic system.

```python
import warnings
warnings.filterwarnings("ignore")
import numpy as np, pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
from sklearn.model_selection import train_test_split, StratifiedKFold, cross_val_score
from sklearn.preprocessing import StandardScaler
from sklearn.pipeline import Pipeline
from sklearn.linear_model import LogisticRegression
from sklearn.neighbors import KNeighborsClassifier
from sklearn.svm import SVC
from sklearn.ensemble import RandomForestClassifier, GradientBoostingClassifier
from sklearn.metrics import accuracy_score, precision_score, recall_score, f1_score, roc_auc_score, confusion_matrix, classification_report, RocCurveDisplay
from sklearn.base import BaseEstimator, TransformerMixin
import joblib
```

```python
df=pd.read_csv("data.csv")
print("Shape:",df.shape)
display(df.head())
print("\nData types:")
display(df.dtypes)
print("\nMissing values:")
display(df.isna().sum().sort_values(ascending=False).head())
print("\nDuplicate rows:",df.duplicated().sum())
```

## Data preprocessing

```python
# Remove identifier and completely empty column
df=df.drop(columns=["Unnamed: 32","id"])
df["diagnosis"]=df["diagnosis"].map({"B":0,"M":1})
display(df.describe().T)
```

## Exploratory Data Analysis

```python
sns.countplot(data=df,x="diagnosis")
plt.xticks([0,1],["Benign","Malignant"])
plt.title("Target Distribution")
plt.show()
print(df["diagnosis"].value_counts())
print(df["diagnosis"].value_counts(normalize=True).mul(100).round(2))
```

```python
df.hist(figsize=(18,16),bins=25)
plt.tight_layout()
plt.show()
```

```python
plt.figure(figsize=(18,14))
sns.heatmap(df.corr(numeric_only=True),cmap="coolwarm",center=0)
plt.title("Correlation Heatmap")
plt.tight_layout()
plt.show()
```

```python
selected=["radius_mean","texture_mean","perimeter_mean","area_mean","concavity_mean","concave points_mean","radius_worst","area_worst"]
fig,axes=plt.subplots(2,4,figsize=(18,8))
for ax,col in zip(axes.ravel(),selected):
    sns.boxplot(data=df,x="diagnosis",y=col,ax=ax)
    ax.set_xticklabels(["Benign","Malignant"])
    ax.set_title(col)
plt.tight_layout(); plt.show()
```

## Outlier analysis
Outliers are inspected but not automatically removed because extreme observations may contain genuine signal.

```python
X0=df.drop(columns="diagnosis")
counts={}
for c in X0.columns:
    q1,q3=X0[c].quantile([.25,.75]); iqr=q3-q1
    counts[c]=((X0[c]<q1-1.5*iqr)|(X0[c]>q3+1.5*iqr)).sum()
display(pd.Series(counts).sort_values(ascending=False).head(15))
```

## Feature engineering

```python
class FeatureEngineer(BaseEstimator,TransformerMixin):
    def fit(self,X,y=None): return self
    def transform(self,X):
        X=X.copy(); e=1e-9
        X["area_radius_ratio"]=X["area_mean"]/(X["radius_mean"]+e)
        X["perimeter_radius_ratio"]=X["perimeter_mean"]/(X["radius_mean"]+e)
        X["worst_mean_radius_ratio"]=X["radius_worst"]/(X["radius_mean"]+e)
        X["worst_mean_area_ratio"]=X["area_worst"]/(X["area_mean"]+e)
        return X
X=df.drop(columns="diagnosis"); y=df["diagnosis"]
X_train,X_test,y_train,y_test=train_test_split(X,y,test_size=.2,stratify=y,random_state=42)
print(X_train.shape,X_test.shape)
```

## Model building and evaluation

```python
models={
"Logistic Regression":Pipeline([("feature_engineering",FeatureEngineer()),("scaler",StandardScaler()),("model",LogisticRegression(max_iter=5000,random_state=42))]),
"KNN":Pipeline([("scaler",StandardScaler()),("model",KNeighborsClassifier(n_neighbors=7))]),
"SVM":Pipeline([("scaler",StandardScaler()),("model",SVC(kernel="rbf",probability=True,random_state=42))]),
"Random Forest":RandomForestClassifier(n_estimators=400,class_weight="balanced",random_state=42,n_jobs=-1),
"Gradient Boosting":GradientBoostingClassifier(random_state=42)}
cv=StratifiedKFold(5,shuffle=True,random_state=42)
rows=[]
for name,m in models.items():
    cvacc=cross_val_score(m,X_train,y_train,cv=cv,scoring="accuracy").mean()
    m.fit(X_train,y_train); p=m.predict(X_test); pr=m.predict_proba(X_test)[:,1]
    rows.append([name,cvacc,accuracy_score(y_test,p),precision_score(y_test,p),recall_score(y_test,p),f1_score(y_test,p),roc_auc_score(y_test,pr)])
results=pd.DataFrame(rows,columns=["Model","CV Accuracy","Test Accuracy","Precision","Recall","F1","ROC-AUC"])
display(results.round(4))
```

## Final model

```python
final_model=models["Logistic Regression"]
pred=final_model.predict(X_test); prob=final_model.predict_proba(X_test)[:,1]
print(classification_report(y_test,pred,target_names=["Benign","Malignant"]))
cm=confusion_matrix(y_test,pred)
sns.heatmap(cm,annot=True,fmt="d",cmap="Blues",xticklabels=["Benign","Malignant"],yticklabels=["Benign","Malignant"])
plt.xlabel("Predicted"); plt.ylabel("Actual"); plt.title("Confusion Matrix"); plt.show()
RocCurveDisplay.from_predictions(y_test,prob)
plt.title("ROC Curve — Final Model"); plt.show()
joblib.dump(final_model,"breast_cancer_model.joblib")
print("Model saved.")
```
